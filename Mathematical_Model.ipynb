## Objective Function

$$
\min Z = \sum_{i \in N, i != j} \sum_{j \in N, i != j} x_{ijw} \cdot d_{ij}
$$

### Decision Variable
$x$ is a given location coordinates from a start location $i \in N$, and a given end location $j \in N$. The race will be require to be completed for a given week throughout the calendar year, best represented by $w \in W$. This will also be a binary variable as described later in the constraints, but, most importantly, there are only 23 races throughout the year; not one for every week of the year.

### Parameters
$d$ is a given distance between coordinates from a start location $i \in N$, and a given end location $j \in N$. 

$u$ is introduced as a numerical order from a start location $i \in N$, and a given end location $j \in N$.

$temp$ is a given average temperature range for a given potential race week throughout the year, best represented by $w \in W$. It is denoted as an expectation as the temperature for a given week is provided by using the historical average for the time horizon given in the datasets.

### Travel Constraints

$$
\sum_{i \in N, i != j} x_{ij} = 1
$$                 

$$
\sum_{j \in N, i != j} x_{ij} = 1
$$                 

** Above two constraints to represent one arrival and one departure location.

$$
x_{ij} \in \{0,1\}
$$

**Above is to denote that the decision variable is boolean.

$$
u_{i}-u_{j}+1 \le 0
$$

**Above to introduce MTZ (Miller–Tucker–Zemlin) subtour elimination.

### Race Week / Climate Constraints

$$
\mu_{t}(w) + 1.645\,\sigma_{t}(w) \le 31
$$

** Above is to denote that races cannot be above 31 using the 95% percentile of historical weekly temperature's normal distriubtion, noted as the current FIA's safety concern.

$$
\mu_{t}(w) + 1.645\,\sigma_{t}(w) \ge 10
$$

** Above is to denote that races cannot be below 10 using the 95% percentile of historical weekly temperature's normal distriubtion, noted as the current FIA's tire handling concern.

$$
\sum_{w \in W} x_{w} = 23 \qquad 
$$  

** Above is to represent that only 23 of the weeks are to be chosen as race weeks throughout the entire calendar year.

$$x_{w} \in \{0,1\} \qquad                  \forall w \in W
$$

** Above is to denote that the decision variable is boolean.
