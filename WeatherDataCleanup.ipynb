```python
import pandas as pd
import glob 
from pandasql import sqldf
```

```python

def merge_weather_data(path):
    files = glob.glob(path + "/*.csv")
    df_list = [pd.read_csv(f) for f in files]
    return pd.concat(df_list, ignore_index=True)

path = "C:/Users/rjasm/OneDrive/Desktop/Projects/Weather_Data"
merged_df = merge_weather_data(path)
```

```python
query = """
SELECT 
    City,
    COUNT(*) AS Not_Null_TAVG,
    SUM(CASE WHEN TAVG IS NULL THEN 1 ELSE 0 END) AS Null_TAVG,
    SUM(CASE WHEN TAVG IS NULL THEN 1 ELSE 0 END)*1.0 / COUNT(*) AS Problematic_TAVG
FROM merged_df
GROUP BY City
ORDER BY Problematic_TAVG DESC;
"""

result = sqldf(query)
result
```
