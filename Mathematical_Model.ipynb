## Objective Function

$$
\min Z = \sum_{i \in N, i != j} \sum_{j \in N, i != j}\sum_{w \in W} x_{ij} \cdot d_{ij} \cdot y_{iw}
$$

### Decision Variable
$x$ is a given location coordinates from a start location $i \in N$, and a given end location $j \in N$.

### Parameters
$d$ is a given distance between coordinates from a start location $i \in N$, and a given end location $j \in N$. 

$u$ is introduced as a numerical order from a start location $i \in N$, and a given end location $j \in N$.

$y$ is introduced as a boolean operator to determine a given race week throughout the calendar year, best represented by $w \in W$ for a given location $i \in N$. Noting that there are only 23 races throughout the year; not one for every week of the year.

$temp$ is a given average temperature range for a given potential race week throughout the year, best represented by $w \in W$ for a given week in the year as well as the location given by $i \in N$. It is used alongside a historical distribution to ensure enhanced adherence to safety protocols in comparison to using an expectation derived from the sample mean.

### Travel Constraints


**Departure Constraint:** Ensures there is a destination location.
$$
\sum_{i \in N, i != j} x_{ij} = 1
$$                 

**Arrival Constraint:** Ensures there is a departure location.

$$
\sum_{j \in N, i != j} x_{ij} = 1
$$                 

**Boolean Constraint:** Ensures that the decision variable is boolean.

$$
x_{ij} \in \{0,1\} \qquad                  \forall i \in N, j \in N
$$

**Boolean Constraint:** Introduces MTZ (Miller–Tucker–Zemlin) subtour elimination.

$$
u_{i}-u_{j}+1 \le 0
$$



### Race Week / Climate Constraints

**Hot Weather Constraint:** Below is used to denote that races cannot be above 31 using the 95% percentile of historical weekly temperature's normal distriubtion, noted as the current FIA's safety concern.

$$
\mu_{iw}(temp) + 1.645\,\sigma_{iw}(temp) \le 31  \qquad                  \forall w \in W, i \in N
$$

**Cold Weather Constraint:** Below is used to denote that races cannot be above 31 using the 95% percentile of historical weekly temperature's normal distriubtion, noted as the current FIA's safety concern.

$$
\mu_{iw}(temp) + 1.645\,\sigma_{iw}(temp) \ge 10  \qquad                  \forall w \in W, i \in N
$$

**Number of Race Constraint:** Below is to represent that only 23 of the weeks are to be chosen as race weeks throughout the entire calendar year.

$$
\sum_{w \in W} \sum_{i \in N} y_{iw} = 23 \qquad 
$$  

**Boolean Race Week Constraint:** Below is to denote that the decision variable is boolean. I.e. Only one race can happen on a given week with a given location.

$$y_{iw} \in \{0,1\} \qquad                  \forall w \in W, i \in N
$$
